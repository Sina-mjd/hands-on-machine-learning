```python
import numpy as np
from sklearn import datasets
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import LinearSVC
```

```python
iris = datasets.load_iris()
X = iris ["data"] [ : , (2,3)]
y = ( iris ["target"] == 2 ).astype(np.float64)
```

```python
svm_clf = Pipeline ([
                   ( "scale" , StandardScaler()),
                   ( "linear_svc" , LinearSVC( C = 1 , loss = "hinge")),
                  ])
```

```python
svm_clf.fit( X , y )
```

```python
svm_clf.predict([[ 5.5 , 1.7 ]])
```

```python
from sklearn.datasets import make_moons
from sklearn.preprocessing import PolynomialFeatures
```

```python
polynomial_svc_clf = Pipeline ([
                              ( "poly_features" , PolynomialFeatures( degree = 3)),
                              ( "scale" , StandardScaler()),
                              ( "linear_svc" , LinearSVC( C = 10 , loss = "hinge")),
                              ]) 
```

```python
polynomial_svc_clf.fit( X , y)
```

```python
X_new = [[1.5, 0.5]]
polynomial_svc_clf.predict ( X_new )
```

```python
import matplotlib.pyplot as plt

plt.plot(X[:, 0][y==0], X[:, 1][y==0], "bs", label="Class 0")
plt.plot(X[:, 0][y==1], X[:, 1][y==1], "g^", label="Class 1")

plt.legend()
plt.show()
```

```python
from sklearn.svm import SVC

poly_kernel_svm_clf = Pipeline([
                              ("scaler", StandardScaler()),
                              ("svm_clf", SVC(kernel="poly", degree=3, coef0=1, C=5))
                              ])
poly_kernel_svm_clf.fit(X, y)
```

```python
X_new = [[ 7 , 2 ]]
poly_kernel_svm_clf.predict (X_new)
```

```python
rbf_kernel_svm_clf = Pipeline([
                             ("scaler", StandardScaler()),
                             ("svm_clf", SVC(kernel="rbf", gamma=5, C=0.001))
                             ])
```

```python
rbf_kernel_svm_clf.fit(X, y)
```

```python
X_new = [[ 7 , 2 ]]
rbf_kernel_svm_clf.predict (X_new)
```
